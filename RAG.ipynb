```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_chroma import Chroma
from langchain_openai import OpenAIEmbeddings
from langchain_openai import ChatOpenAI
from langchain_core.documents import Document
```

```python
from openai import OpenAI
client = OpenAI()
response = client.responses.create(model = "gpt-4.1-mini",
                                  input = "Will Character CC cream increase pimples?")
print(response.output_text)
```

```python
documents = Document(page_content = "Students need 80% attendance anyhow.")
splitter = RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=50)
chunks = splitter.split_documents(documents)
```

```python
embeddings = OpenAIEmbeddings()
database = Chroma.from_documents(chunks,embeddings)
```

```python
user_question = "How much attendance is required?"
results = database.similarity_search(user_question,k=2)
```

```python
context = "\n".join(doc.page_content for doc in results)
```

```python
llm = ChatOpenAI(model="gpt-4.1-mini")
answer = llm.invoke(f"Answer this question : {user_question} by using only this context: {context}")
print(answer.content)
```
